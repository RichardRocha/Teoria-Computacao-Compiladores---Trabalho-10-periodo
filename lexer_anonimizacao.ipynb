```python
from ply.lex import lex

tokens = ( 'CPF', 'EMAIL', 'TELEFONE', 'CEP', 'DATA', 'NUMERO', 'PALAVRA', 'PONTUACAO' )

def t_CPF(t):
    r'\d{3}\.\d{3}\.\d{3}\-\d{2}'
    return t

def t_TELEFONE(t):
    r'\(\d{2}\)\s\d{5}\-\d{4}'
    return t

def t_CEP(t):
    r'\d{5}\-\d{3}'
    return t

def t_EMAIL(t):
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
    return t

def t_DATA(t):
    r'\d{2}\/\d{2}\/\d{4}'
    return t

def t_NUMERO(t):
    r'\b\d+(?:[.,]\d+)?\b'
    return t

t_PALAVRA = r"[A-Za-zÀ-ÿ]+(?:[-'][A-Za-zÀ-ÿ]+)*"
t_PONTUACAO = r'[.,!?;:()]'
```

```python
t_ignore = " \t\r"
```

```python
def t_newline(t):
    r"\n+"
    t.lexer.lineno += len(t.value)
```

```python
def t_error(t):
    print(f"Lexema Invalido: {t.value} na linha {t.lineno} e coluna {t.lexpos}")
    t.lexer.skip(1)
    return t
```

```python
import os
import sys
import types
from ply.lex import lex
from ply.yacc import yacc

# Workaround for ply in Colab/Jupyter environments where __main__ module lacks __file__
if '__main__' in sys.modules and not hasattr(sys.modules['__main__'], '__file__'):
    sys.modules['__main__'].__file__ = '__temp_colab_module__.py'

# Remover arquivos temporários do lextab
for f in ['lextab.py', 'lextab.pyc', 'parsetab.py', 'parsetab.pyc']:
    if os.path.exists(f):
        try:
            os.remove(f)
        except Exception:
            pass

# Limpar do cache do sys.modules
for m in ['lextab', 'parsetab']:
    if m in sys.modules:
        del sys.modules[m]

# Limpar apenas variáveis de string de tokens não especificados (mantendo funções como t_error e t_newline)
current_tokens = set(globals().get('tokens', []))
for key in list(globals().keys()):
    if key.startswith('t_') and key != 't_ignore':
        # Não deletar funções legítimas do lexer (como t_error, t_newline, etc)
        if isinstance(globals()[key], (types.FunctionType, types.MethodType)):
            continue
        token_name = key[2:]
        if token_name not in current_tokens:
            del globals()[key]
```

```python
lexer = lex(object=sys.modules[__name__], debug=False, optimize=False, lextab='lextab')
```
